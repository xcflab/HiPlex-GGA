```python
import pandas as pd
batch_1 = pd.read_csv("/xcfhome/zewu/ngs/N2422148_80-1626558504_2024-09-11/240909-A00599B/count_batch1.csv", header=0)
batch_2 = pd.read_csv("/xcfhome/zewu/ngs/N2422148_80-1626558504_2024-09-11/240909-A00599B/count_batch2.csv", header=0)
batch_3 = pd.read_csv("/xcfhome/zewu/ngs/N2422148_80-1626558504_2024-09-11/240909-A00599B/count_batch3.csv", header=0)
```

```python

import matplotlib.pyplot as plt


# 使用 pandas 读取文件
df1 = batch_1
df2 = batch_2
df3 = batch_3

# 合并三个数据集
#df = pd.concat([df1, df2, df3])
df = df3
# 确保第二列是样品的数量，并将其命名为 'count'
df.columns = ['UMI', 'Reads']  # 假设样品名在第一列，数量在第二列

# 按样品数量统计样品的个数
count_distribution = df['Reads'].value_counts().sort_index()

# 生成柱状图
plt.figure(figsize=(4, 3))
plt.scatter(count_distribution.index, count_distribution.values, edgecolors='black', facecolors='none')

plt.yscale('log')
plt.ylim(1, 130000)
plt.xlim(0, 1000)
# 设置图表标题和轴标签
plt.title('Sample Count Distribution', fontsize=16)
plt.xlabel('Sample Length', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)

# 显示图表
plt.show()
```
